```python
#Highest order day/date
Order_query4 = """
SELECT
    CASE strftime('%w', order_date)
        WHEN '0' THEN 'Sunday'
        WHEN '1' THEN 'Monday'
        WHEN '2' THEN 'Tuesday'
        WHEN '3' THEN 'Wednesday'
        WHEN '4' THEN 'Thursday'
        WHEN '5' THEN 'Friday'
        WHEN '6' THEN 'Saturday'
    END AS day_name,
    COUNT(*) AS total_count
FROM Order_details
GROUP BY day_name
ORDER BY total_count DESC
"""
Order_q4 = pd.read_sql(Order_query4, conn)
Order_q4
```

```python
day_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

Order_q4["day_name"] = pd.Categorical(
    Order_q4["day_name"],
    categories=day_order,
    ordered=True
)

Order_q4 = Order_q4.sort_values("day_name")
fig_o4 = px.line(
    Order_q4,
    x="day_name",
    y="total_count",
    markers=True,
    title="Orders by Day of Week"
)
fig_o4
```

```python
#Which day of the week generates the highest revenue?
Order_query5 = """
SELECT
    CASE strftime('%w', order_date)
        WHEN '0' THEN 'Sunday'
        WHEN '1' THEN 'Monday'
        WHEN '2' THEN 'Tuesday'
        WHEN '3' THEN 'Wednesday'
        WHEN '4' THEN 'Thursday'
        WHEN '5' THEN 'Friday'
        WHEN '6' THEN 'Saturday'
    END AS day_name,sum(order_value) as total_value
    
FROM Order_details
GROUP BY day_name
ORDER BY total_value DESC
"""
Order_q5 = pd.read_sql(Order_query5, conn)
Order_q5
```
